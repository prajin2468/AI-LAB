```python
import heapq


GOAL_STATE = (1, 2, 3,
              4, 5, 6,
              7, 8, 0)

class Node:
    def __init__(self, state, parent, move, g, h):
        self.state = state
        self.parent = parent
        self.move = move
        self.g = g
        self.h = h
        self.f = g + h

    def __lt__(self, other):
        return self.f < other.f

def calculate_manhattan(state):
    distance = 0
    for idx, value in enumerate(state):
        if value != 0:

            target_row = (value - 1) // 3
            target_col = (value - 1) % 3

            curr_row = idx // 3
            curr_col = idx % 3
            distance += abs(curr_row - target_row) + abs(curr_col - target_col)
    return distance

def get_neighbors(node):
    neighbors = []
    idx = node.state.index(0)
    row, col = idx // 3, idx % 3


    directions = [
        (-1, 0, 'Up'),
        (1, 0, 'Down'),
        (0, -1, 'Left'),
        (0, 1, 'Right')
    ]

    for dr, dc, move in directions:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_idx = new_row * 3 + new_col

            state_list = list(node.state)
            state_list[idx], state_list[new_idx] = state_list[new_idx], state_list[idx]
            new_state = tuple(state_list)


            g = node.g + 1
            h = calculate_manhattan(new_state)
            neighbors.append(Node(new_state, node, move, g, h))

    return neighbors

def solve_8_puzzle(start_state):
    open_set = []
    closed_set = set()

    start_node = Node(start_state, None, None, 0, calculate_manhattan(start_state))
    heapq.heappush(open_set, start_node)

    while open_set:
        current = heapq.heappop(open_set)

        if current.state == GOAL_STATE:
            path = []
            while current.parent:
                path.append((current.move, current.state))
                current = current.parent
            return path[::-1]
        closed_set.add(current.state)

        for neighbor in get_neighbors(current):
            if neighbor.state in closed_set:
                continue
            heapq.heappush(open_set, neighbor)

    return None

def print_board(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


if __name__ == "__main__":

    initial_state = (1, 2, 3,
                     4, 0, 5,
                     7, 8, 6)

    print("Initial State:")
    print_board(initial_state)

    solution = solve_8_puzzle(initial_state)

    if solution:
        print(f"Solution found in {len(solution)} moves!\n")
        for step, state in solution:
            print(f"Move: {step}")
            print_board(state)
    else:
        print("No solution exists for this configuration.")
```
